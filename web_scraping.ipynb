```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.9; rv:50.0) Gecko/20100101 Firefox/50.0"}
webpage = requests.get("https://webscraper.io/test-sites/pagination", headers = headers).text
```

```python
soup = BeautifulSoup(webpage, 'lxml')
```

```python
print(soup.prettify)
```

```python

for i in soup.find_all('p', class_='price col-6 mb-0 text-end p-0'):
    print(i.text.strip())
```

```python
car = soup.find_all('div', class_='col-md-4 col-xl-4 col-lg-4')
```

```python
len(car)
```

```python
from pandas._config.config import describe_option
name = []
description = []
for i in car:
  name.append(i.find('h3').text.strip())
  description.append(i.find_all('p', class_='description text-muted')[0].text)
```
